```python
import numpy as np
import pandas as pd
rng = np.random.default_rng(42)
n = 500
amount = rng.lognormal(mean=5.0,sigma=1.0,size=n)
hour = rng.integers(0,24,n)
new_device = rng.binomial(1,0.25,n)
foreign_ip = rng.binomial(1,0.12,n)
score = -4.5 + 0.45*(amount > 300) + 1.1*new_device+1.5*foreign_ip + 0.7*(hour < 5)
prob = 1/(1+np.exp(-score))
isFraud = rng.binomial(1,prob)
df = pd.DataFrame({'amount':amount.round(2),'hour':hour,'new_device':new_device,'foreign_ip':foreign_ip,'isFraud':isFraud})

print(df.head())
print('Distribusi kelas:')
print(df['isFraud'].value_counts())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier

X = df.drop(columns='isFraud')
y = df['isFraud']
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.30, random_state=42,stratify=y)
model = DecisionTreeClassifier(max_depth=4,class_weight='balanced',random_state=42)
model.fit(X_train,y_train)
y_pred = model.predict(X_test)
print('Data uji',len(y_test))
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test,y_pred,labels=[0,1])
tn,fp,fn,tp = cm.ravel()
print('Confusion matrix:\n',cm)
print(f'TP={tp},TN={tn},FP={fp},FN={fn}')
```

```python
accuracy = (tp + tn) / (tp + tn + fp + fn)
precision = tp / (tp + fp) if tp + fp else 0
recall = tp / (tp + fn) if tp + fn else 0

f1 = 2*precision*recall/(precision+recall) if precision+recall else 0
print(f'accuracy={accuracy:.3f};precision={precision:.3f};Recall={recall:.3f};F1={f1:.3f}')
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score
print(accuracy_score(y_test,y_pred),precision_score(y_test,y_pred,zero_division=0),
      recall_score(y_test,y_pred,zero_division=0),f1_score(y_test,y_pred,zero_division=0))
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay
ConfusionMatrixDisplay(confusion_matrix=cm,
display_labels=['Normal','Fraud']).plot(cmap='Blues', values_format='d')
plt.title('Confusion Matrix – Deteksi Fraud')
plt.tight_layout()
plt.show()
```

```python
import hashlib

def hmac_manual(key, message):
    block_size = 64

    # Menyesuaikan panjang key
    if len(key) > block_size:
        key = hashlib.sha256(key).digest()

    key = key.ljust(block_size, b'\x00')

    # Membuat ipad dan opad
    ipad = bytes([0x36] * block_size)
    opad = bytes([0x5c] * block_size)

    # XOR key dengan padding
    inner_key = bytes(a ^ b for a, b in zip(key, ipad))
    outer_key = bytes(a ^ b for a, b in zip(key, opad))

    # Inner hash
    inner_hash = hashlib.sha256(
        inner_key + message
    ).digest()

    # Outer hash
    result = hashlib.sha256(
        outer_key + inner_hash
    ).hexdigest()

    return result


key = b"kunci-rahasia"
message = b"Belajar Kriptografi"

print(hmac_manual(key, message))
```
